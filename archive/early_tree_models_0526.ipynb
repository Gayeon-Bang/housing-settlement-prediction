미충족 시 → 거의 대부분 수도권 내 이주자

충족 시 → 다음 분기 필요

직업대분류_기타 여부

산업대분류 + 경제활동 상태

제조업 & 비경제활동 상태일 경우 → 수도권 외 유입자 가능성 상승

| 기준            | 비수도권 → 수도권 이주자 가능성 ↑ |
| ------------- | -------------------- |
| 만연령 ≤ 23세     | 청년층일수록               |

| 미혼 상태         | 기혼보다 외부 유입 가능성 ↑     |
| 사무/전문직        | 수도권 전문직 유입 수요 반영     |
| 통근시간 ≥ 30분 이상 | 장거리 통근 감수 이주 가능성     |
| 제조업/경제비활동     | 외부 유입 후 비활동 상태 존재    |


```python
# Re-import all necessary libraries after code reset
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Reload the dataset
file_path = "인구주택총조사_2%_인구사항(제공)_2020.csv"
df = pd.read_csv(file_path, encoding='cp949')

# 재정의: 수도권 분류 및 이동유형
sudogwon = ['서울특별시', '경기도', '인천광역시']
def classify_movement(row):
    curr = row['행정구역시도코드_LABEL']
    past = row['1년전거주지_다른시도_LABEL']
    if pd.isna(past) or past == curr:
        return '수도권 내 이동' if curr in sudogwon else '기타'
    elif past not in sudogwon and curr in sudogwon:
        return '비수도권 → 수도권'
    elif past in sudogwon and curr in sudogwon:
        return '수도권 내 이동'
    else:
        return '기타'
df['이동유형'] = df.apply(classify_movement, axis=1)
df['이동타겟'] = df['이동유형'].apply(lambda x: 1 if x == '비수도권 → 수도권' else (0 if x == '수도권 내 이동' else None))
df_model = df[df['이동타겟'].notnull()].copy()

# 교통수단 변수
transport_columns = [
    '이용교통수단-걸어서_LABEL',
    '이용교통수단-자전거_LABEL',
    '이용교통수단-승용차^ 승합차_LABEL',
    '이용교통수단-트럭_LABEL',
    '이용교통수단-시내^ 좌석^ 마을버스_LABEL',
    '이용교통수단-통근 통학차량_LABEL',
    '이용교통수단-고속^ 시외버스_LABEL',
    '이용교통수단-전철^ 지하철_LABEL',
    '이용교통수단-기차_LABEL',
    '이용교통수단-택시_LABEL',
    '이용교통수단-기타(오토바이^ 배 등)_LABEL'
]

# 클러스터링용 변수
features_cluster = [
    '만연령', '통근통학_소요분수',
    '혼인상태_LABEL', '교육정도_LABEL', '직업대분류_LABEL',
    '산업대분류_LABEL', '경제활동상태_LABEL'
] + transport_columns

# 전처리
df_cluster = df_model[features_cluster].copy()
df_cluster = pd.get_dummies(df_cluster, drop_first=True)
df_cluster = df_cluster.dropna()

# 표준화
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_cluster)

# 차원 축소
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

# KMeans 클러스터링
kmeans = KMeans(n_clusters=4, random_state=42)
clusters = kmeans.fit_predict(X_scaled)

# 시각화 준비
cluster_df = pd.DataFrame(X_pca, columns=['PCA1', 'PCA2'])
cluster_df['Cluster'] = clusters

# 시각화
plt.figure(figsize=(10, 6))
for cluster_id in cluster_df['Cluster'].unique():
    cluster_data = cluster_df[cluster_df['Cluster'] == cluster_id]
    plt.scatter(cluster_data['PCA1'], cluster_data['PCA2'], label=f'Cluster {cluster_id}', alpha=0.6)

plt.title('이주자 유형 클러스터링 (PCA 2D)')
plt.xlabel('PCA 1')
plt.ylabel('PCA 2')
plt.legend()
plt.grid(True)
plt.show()
```
